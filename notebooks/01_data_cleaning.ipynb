```python
import pandas as pd
df = pd.read_csv("../data/raw/essai_clinique.csv", sep=";", skiprows=1)
```

```python
df = df.dropna(how="all")
df["Dose (mg)"] = pd.to_numeric(df["Dose (mg)"], errors="coerce")
df["Biomarqueur J28"] = df["Biomarqueur J28"].str.replace(",", ".")
df["Biomarqueur J28"] = pd.to_numeric(df["Biomarqueur J28"])
df["Biomarqueur J0"] = df["Biomarqueur J0"].str.replace(",", ".")
df["Biomarqueur J0"] = pd.to_numeric(df["Biomarqueur J0"])
df["Poids (kg)"] = df["Poids (kg)"].str.replace(",", ".").str.replace(" kg", "")
df["Poids (kg)"] = pd.to_numeric(df["Poids (kg)"])
df["Age (ans)"] = df["Age (ans)"].str.replace(" ans", "")
df["Age (ans)"] = pd.to_numeric(df["Age (ans)"], errors="coerce")
df["Duree traitement"] = df["Duree traitement"].str.replace(" jours", "")
df["Duree traitement"] = pd.to_numeric(df["Duree traitement"])
print(df.head())
print(df.info())
```

```python
print(df[df.duplicated(keep=False)])
df = df.drop_duplicates()
print("duplicated", df.duplicated().sum())
print(df.isna().sum())
```

```python
df=df.drop("Unnamed: 14", axis=1)
print(df.shape)
df[df["Dose (mg)"].isna()]
df[df["Biomarqueur J28"].isna()]
df[df["Effet indesirable"].isna()]
df["Traitement"] = df["Traitement"].str.upper()
df["Traitement"].value_counts()
df["Sexe"]=df["Sexe"].str.replace("Femme", "F").str.replace("Homme", "H")
df["Sexe"]=df["Sexe"].str.upper()
df["Sexe"].value_counts()
df["Effet indesirable"]=df["Effet indesirable"].str.replace(" oui", "Oui").str.replace(" non", "Non")
df["Effet indesirable"].value_counts()
df["Date visite"] = pd.to_datetime(df["Date visite"], format="%d/%m/%Y")
#print("describe", df.describe())
print("type", df.dtypes)
df = df.drop("Commentaire", axis=1)
df.to_csv("../data/processed/essai_clinique_clean.csv", sep=";", index=False)
```
